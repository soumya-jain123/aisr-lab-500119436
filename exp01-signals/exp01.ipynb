```python
import numpy as np
import matplotlib.pyplot as plt

g = 9.81
L = 1.0
theta_0 = np.radians(50)

dt = 0.01
t = np.arange(0, 10, dt)

omega_n = np.sqrt(g / L)
theta_t = theta_0 * np.cos(omega_n * t)

N = len(t)
fft_vals = np.fft.fft(theta_t)
freqs = np.fft.fftfreq(N, dt)

pos_mask = freqs >= 0
freqs = freqs[pos_mask]
fft_magnitude = np.abs(fft_vals[pos_mask]) * (2.0 / N)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(t, np.degrees(theta_t), color="crimson")
plt.title("Pendulum Simulation - Time Domain")
plt.xlabel("Time (s)")
plt.ylabel("Angle (deg)")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.plot(freqs, fft_magnitude, color="purple")
plt.title("Frequency Domain (FFT Spectrum)")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.xlim(0, 2)
plt.grid(True)

plt.tight_layout()
plt.show()
```
